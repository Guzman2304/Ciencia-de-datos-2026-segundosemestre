```python
with open("/content/charlaUlibro.txt","r",encoding="utf-8") as archivo:
  charla=archivo.read()
```

```python
charla=charla.lower()
charla=charla.replace(",","")
charla=charla.replace(";","")
charla=charla.replace(".","")
charla=charla.replace("?","")
charla=charla.replace("¿","")
charla=charla.replace("¡","")
charla=charla.replace("!","")
charla=charla.replace("—","")
charla=charla.replace(":","")
charla=charla.replace('“',"")
palabras=charla.split()
```

```python
import pandas as pd
import numpy as np

palabras_array=np.array(palabras)
longitudes=np.array([len(palabra) for palabra in palabras_array])

df=pd.DataFrame(
    {
        "palabra":palabras,
        "longitud":longitudes
    }
)
```

```python
stopwords = [
    "el","que","en","de","del","y","la","a","los","lo","al","muy","un","te","es","se","no","si","le","su",'mi',"por","o","son","eso","con","para","una","más","soy","así","las"
    ,"este","él","tan","con","esa","qué","toc","mi","pues","quien","muy","allá","porque","me","pero","uno","como","ese",
    "nos","he","sea","hay","sí","etcétera","creo","yo","va","mucho","cada","sobre","entonces","ya","muchas","veces","ni","está","través",
    "cuando","ustedes"
]
```

```python
df_limpio=df[~df["palabra"].isin(stopwords)]
frecuencias=df_limpio["palabra"].value_counts()
frecuencias.head(20)
```

```python
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from PIL import Image

texto_limpio=" ".join(df_limpio["palabra"])

nube=WordCloud(
  width=1200,
  height=600,
  background_color="white",
  colormap="Oranges_r"
).generate(texto_limpio)

plt.figure(figsize=(15,7))
plt.imshow(nube)
plt.axis("off")
plt.show()
```
